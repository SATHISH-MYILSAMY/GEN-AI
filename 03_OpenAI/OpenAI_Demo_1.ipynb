### ChatCompletion API and Completion API

#### 1) Completion API: Hooks you up with text completions from a single prompt

#### 2) Chat Completion API: Nails it in the chat game, keeping the conversational flow intract

### ChatCompletion API

```python
propmt = "What is the capital of France?"

response = openai.ChatCompletion.create(
    model="gpt-5.5",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": propmt}
    ]
        
)

print(response)
```

```python
propmt = "What is the capital of France?"

response = openai.ChatCompletion.create(
    model="gpt-5.5",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": propmt}
    ]
        
)

print(response["choices"][0]["message"]["content"])
```

```python
## ChatCompletion can understand multiple prompting in a single request. It can also understand the context of the conversation. 

propmt1 = "Hello, How are you?"
propmt2 = "I'm a 25 years old & I'm a Software Engineer"
propmt3 = "Tell me about me"

response = openai.ChatCompletion.create(
    model="gpt-5.5",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": propmt1},
        {"role": "user", "content": propmt2},
        {"role": "user", "content": propmt3}
    ]
        
)

print(response["choices"][0]["message"]["content"])
```

### Completion API

```python
prompt = "What is Python?"

response = openai.Completion.create(
    model = "babbage-002",
    prompt = prompt
)

print(response)
```

```python
prompt = "What is Python?"

response = openai.Completion.create(
    model="babbage-002",
    prompt=prompt,
    max_tokens=50
)

print(response["choices"][0]["text"])
```
